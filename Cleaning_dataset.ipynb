```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/capstone/Skincare Treatment Dataset.csv')
print("Shape awal:", df.shape)
df.head()
```

```python
data = df.copy()

data.columns = (
  data.columns
  .str.strip()
  .str.lower() #mengecilkan huruf
  )

data.head()
```

```python
print("Jumlah duplikat:", data.duplicated().sum())
data = data.drop_duplicates().reset_index(drop=True)
print("Shape setelah hapus duplikat:", data.shape)
```

```python
data.isnull().sum()
```

```python
# Inisialisasi data dari df jika belum ada, dan pastikan kolom menggunakan huruf kecil
import pandas as pd


data = df.copy()
data.columns = data.columns.str.strip().str.lower()

kolom_kategorik = [
    'skin_type',
    'skin_subtype',
    'sensitivity',
    'concern',
    'internal_type'
]

for kolom in kolom_kategorik:
    data[kolom] = data[kolom].astype(str).str.lower().str.strip()

# Memeriksa distribusi kategori
data['skin_type'].value_counts()
```

```python
df.head()
```

```python
label_skin = {
    'normal': 0,
    'dry': 1,
    'oily': 2,
    'combination': 3
}
data['skin_type'] = data['skin_type'].map(label_skin)

# cek: harus 0, kalau tidak berarti ada nilai yang tidak cocok dengan kamus
print(data['skin_type'].isna().sum())
```

```python
data
```

```python
label_sensitivity = {'no': 0, 'yes': 1}
data['sensitivity'] = data['sensitivity'].map(label_sensitivity)

# cek: harus 0, kalau tidak berarti ada nilai yang tidak cocok dengan kamus
print(data['sensitivity'].isna().sum())
print(data['sensitivity'].value_counts())
```
